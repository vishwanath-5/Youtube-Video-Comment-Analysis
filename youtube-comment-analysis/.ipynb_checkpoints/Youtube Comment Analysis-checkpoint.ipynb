```python
f = open('stop_hinglish.txt', 'r')
stop_words = f.read()
print(stop_words)
```

```python
words = []
for message in df['message']:
    for word in message.lower().split():
        if word not in stop_words:
            words.append(word)
```

```python
from collections import Counter
pd.DataFrame(Counter(words).most_common(20))
```
